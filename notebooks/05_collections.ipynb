Lists, dictionaries and tuples cover most needs. But some jobs come up so often that Python ships a faster, cleaner tool for each one. The collections module gives you four favourites. Counter counts things. defaultdict groups things without key errors. deque is a fast queue. namedtuple is a tiny record with named fields.

**In this notebook:** You will find ShopEasy's best selling products, group orders by city, keep a rolling list of the last three page visits, and make readable order records.

## Step 1 · Counter: count anything
Pass any list. `most_common(n)` gives the top n.

````python
from collections import Counter, defaultdict, deque, namedtuple

sold = ["bottle", "mat", "bottle", "lamp", "bottle", "mat", "pen"]
counts = Counter(sold)
print(counts)
print("Top 2:", counts.most_common(2))
print("Lamps sold:", counts["lamp"], "| Chairs sold:", counts["chair"])
````

## Step 2 · Counters add up
You can add, subtract and update counters, which is handy for daily totals.

````python
monday = Counter(bottle=3, mat=1)
tuesday = Counter(bottle=1, lamp=2)
print("Two days:", monday + tuesday)
print("Letters in 'banana':", Counter("banana").most_common())
````

## Step 3 · defaultdict: group without errors
`defaultdict(list)` creates an empty list the first time you use a new key.

````python
orders = [("Pune", 499), ("Delhi", 1299), ("Pune", 899), ("Mumbai", 199), ("Delhi", 499)]
by_city = defaultdict(list)
for city, amount in orders:
    by_city[city].append(amount)

for city, amounts in by_city.items():
    print(f"{city:<7} {amounts}  total={sum(amounts)}")
````

## Step 4 · deque: fast queue with a limit
`maxlen=3` keeps only the newest three items, which is perfect for a 'recently viewed' list.

````python
recent = deque(maxlen=3)
for page in ["home", "bottle", "mat", "lamp", "cart"]:
    recent.append(page)
    print(list(recent))

queue = deque(["order-1", "order-2"])
queue.append("order-3")
print("Packing:", queue.popleft(), "| left:", list(queue))
````

## Step 5 · namedtuple: readable records
Fields by name instead of `row[2]`. Still light and fast like a tuple.

````python
Order = namedtuple("Order", ["id", "city", "amount"])
o = Order(1001, "Pune", 499)
print(o)
print(o.city, o.amount)
print(o._asdict())
````
