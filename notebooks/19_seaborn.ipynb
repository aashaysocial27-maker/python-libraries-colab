Seaborn sits on top of Matplotlib and speaks pandas. You give it a DataFrame and column names, and it draws a polished statistical chart, with averages, error bars and colour groups worked out for you. What takes ten lines in Matplotlib often takes one in Seaborn.

**In this notebook:** You will compare average order value by category and city, see the full spread of orders with box and violin plots, find which numbers move together with a correlation heatmap, and plot a trend line with its confidence band.

## Step 1 · Create the ShopEasy data
Same dataset as before.

````python
import numpy as np, pandas as pd

rng = np.random.default_rng(42)
n = 600
orders = pd.DataFrame({
    "order_id": np.arange(1001, 1001 + n),
    "date": pd.Timestamp("2026-01-01") + pd.to_timedelta(rng.integers(0, 90, n), unit="D"),
    "city": rng.choice(["Mumbai", "Delhi", "Pune", "Bengaluru"], n),
    "category": rng.choice(["Books", "Electronics", "Clothing", "Home"], n, p=[0.3, 0.2, 0.3, 0.2]),
    "quantity": rng.integers(1, 6, n),
    "price": rng.choice([199, 499, 999, 1499, 2999], n),
    "rating": rng.choice([1, 2, 3, 4, 5], n, p=[0.05, 0.1, 0.2, 0.35, 0.3]),
})
orders["revenue"] = orders["quantity"] * orders["price"]
orders.to_csv("shop_orders.csv", index=False)
print(orders.shape, "-> saved shop_orders.csv")
orders.head()
````

## Step 2 · Average order value with hue
`hue` splits each bar by city. The black lines are 95% confidence intervals.

````python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
plt.figure(figsize=(9, 4))
ax = sns.barplot(data=orders, x="category", y="revenue", hue="city")
ax.set_title("Average order value by category and city")
plt.show()
````

## Step 3 · Distributions: box, violin and histogram
Box plots show median and spread; violins show the shape; `histplot` with `kde=True` adds a smooth curve.

````python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
sns.boxplot(data=orders, x="category", y="revenue", ax=axes[0])
sns.violinplot(data=orders, x="city", y="rating", ax=axes[1], inner="quartile")
sns.histplot(data=orders, x="revenue", hue="category", kde=True, element="step", ax=axes[2])
axes[0].set_title("Order value"); axes[1].set_title("Ratings by city"); axes[2].set_title("Order value spread")
plt.tight_layout(); plt.show()
````

## Step 4 · Correlation heatmap
Which numbers move together? 1 means perfectly together, 0 means unrelated.

````python
corr = orders[["quantity", "price", "revenue", "rating"]].corr()
plt.figure(figsize=(5, 4))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1)
plt.title("Correlation between order numbers")
plt.show()
````

## Step 5 · Trends with regression lines
`lmplot` draws a fitted line with its confidence band for each group.

````python
daily = orders.groupby(["date", "category"], as_index=False)["revenue"].sum()
daily["day"] = (daily["date"] - daily["date"].min()).dt.days
g = sns.lmplot(data=daily, x="day", y="revenue", hue="category", height=4, aspect=1.8, scatter_kws={"alpha": 0.4})
g.set(title="Daily revenue trend by category")
plt.show()
````
