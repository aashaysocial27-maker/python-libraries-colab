## Step 2 · Split into sentences and words
Tokenising handles punctuation and contractions properly.

````python
from nltk.tokenize import sent_tokenize, word_tokenize
reviews = [
    "Absolutely love this steel bottle, keeps water cold all day!",
    "The yoga mat smells bad and started tearing after a week. Very disappointed.",
    "Delivery was quick and the desk lamp is bright. Good value for money.",
    "Terrible customer service. My refund still hasn't arrived after 20 days.",
    "Decent headphones, but the battery life is shorter than advertised.",
    "Best purchase this year! The cotton shirts fit perfectly and feel great.",
    "The book arrived with a torn cover, but the story is wonderful.",
    "Not worth the price. The blender stopped working in two days.",
]

text = " ".join(reviews)
print(len(sent_tokenize(text)), "sentences")
print(word_tokenize(reviews[3]))
````

## Step 3 · Clean: lower case, no punctuation, no stopwords
Keep `not` and `no`: they matter for meaning.

````python
from nltk.corpus import stopwords
stops = set(stopwords.words("english")) - {"not", "no"}

def clean(t):
    return [w for w in word_tokenize(t.lower()) if w.isalpha() and w not in stops]

cleaned = [clean(r) for r in reviews]
print(cleaned[1])
````

## Step 4 · Stemming vs lemmatisation
Stemming chops endings (fast, rough). Lemmatising looks words up (slower, real words).

````python
from nltk.stem import PorterStemmer, WordNetLemmatizer
stem, lemma = PorterStemmer(), WordNetLemmatizer()
for w in ["loving", "studies", "arrived", "better", "batteries"]:
    print(f"{w:<10} stem: {stem.stem(w):<8} lemma(verb): {lemma.lemmatize(w, 'v'):<8} lemma(noun): {lemma.lemmatize(w)}")
````

## Step 5 · What do customers talk about?
`FreqDist` counts words. Also tag parts of speech to keep only nouns.

````python
from nltk import FreqDist, pos_tag
all_words = [lemma.lemmatize(w) for ws in cleaned for w in ws]
print("Top words:", FreqDist(all_words).most_common(8))
nouns = [w for w, tag in pos_tag(all_words) if tag.startswith("NN")]
print("Top nouns:", FreqDist(nouns).most_common(6))
````

## Step 6 · Sentiment with VADER
A rule-based scorer made for short reviews. `compound` runs from -1 (very negative) to +1 (very positive).

````python
from nltk.sentiment import SentimentIntensityAnalyzer
sia = SentimentIntensityAnalyzer()
for r in reviews:
    s = sia.polarity_scores(r)["compound"]
    mood = "positive" if s > 0.05 else "negative" if s < -0.05 else "neutral"
    print(f"{s:+.2f} {mood:<8} {r[:55]}")
````
