NumPy is the foundation of data science in Python. Its main object is the array, a grid of numbers that all share one type. The magic is vectorisation. ## Step 1 · Create arrays
From a list, or with helpers like `arange`, `zeros` and `linspace`.

````python
import numpy as np

prices = np.array([199, 499, 999, 1499, 2999])
print(prices, prices.dtype, prices.shape)
print(np.arange(0, 10, 2))
print(np.zeros(3))
print(np.linspace(0, 1, 5))
````

## Step 2 · Vectorised maths
One expression works on every element. No loop.

````python
with_gst = prices * 1.18
print(with_gst.round(2))
print("Total:", prices.sum(), "| Mean:", prices.mean(), "| Max:", prices.max())

big = np.random.default_rng(0).random(1_000_000)
import time
t = time.perf_counter(); s1 = sum(x * 2 for x in big); loop = time.perf_counter() - t
t = time.perf_counter(); s2 = (big * 2).sum(); vec = time.perf_counter() - t
print(f"Python loop {loop:.3f}s vs NumPy {vec:.4f}s -> about {loop / vec:.0f}x faster")
````

## Step 3 · 2-D arrays and axis
Rows are products, columns are days. `axis=1` totals across each row.

````python
rng = np.random.default_rng(42)
sales = rng.integers(0, 20, size=(4, 7))       # 4 products x 7 days
products = ["Bottle", "Mat", "Lamp", "Pen"]
print(sales)
print("Shape:", sales.shape)
print("Per product (axis=1):", sales.sum(axis=1))
print("Per day     (axis=0):", sales.sum(axis=0))
print("Best day index:", sales.sum(axis=0).argmax())
````

## Step 4 · Indexing, slicing and masks
`[row, col]` picks cells. A comparison gives a True/False mask you can filter with.

````python
print("Mat on day 3:", sales[1, 2])
print("Lamp all week:", sales[2])
print("All products, first 3 days:\n", sales[:, :3])

mask = sales > 15
print("Cells above 15:", sales[mask])
print("Days each product beat 15:", mask.sum(axis=1))
````

## Step 5 · Broadcasting
A shape (4,1) column stretches across a (4,7) table automatically.

````python
unit_price = np.array([[499], [1299], [899], [49]])     # shape (4, 1)
revenue = sales * unit_price                            # shape (4, 7)
print(revenue)
for name, total in zip(products, revenue.sum(axis=1)):
    print(f"{name:<7} Rs.{total:,}")
````

## Step 6 · Reshape and stats
`reshape` changes shape without changing data. `percentile` and `std` describe spread.

````python
a = np.arange(12)
print(a.reshape(3, 4))
daily = sales.sum(axis=0)
print("Std dev:", daily.std().round(2), "| 90th percentile:", np.percentile(daily, 90))
print("Days above average:", np.where(daily > daily.mean())[0])
````
