# 1. Files & Folders — `os · sys · pathlib`

_Find, create, read and organise files from Python._

Almost every real program touches files. It reads a CSV, saves a report, or tidies a downloads folder. Python gives you three built-in tools for this. pathlib treats a file path as an object you can join, inspect and loop over. os talks to the operating system. sys tells you about Python itself.

## Step 1 · Where am I?
`Path.cwd()` is the folder Python is working in. In Colab this is `/content`. `sys` tells you which Python is running.

````python
from pathlib import Path
import os, sys

print("Python version:", sys.version.split()[0])
print("Working folder:", Path.cwd())
print("Home folder:   ", Path.home())
````

## Step 2 · Build paths the safe way
The `/` operator joins path parts. A Path knows its name, suffix and parent folder.

````python
project = Path("shopeasy")
report = project / "reports" / "sales_jan.csv"

print(report)          # shopeasy/reports/sales_jan.csv
print(report.name)     # sales_jan.csv
print(report.stem)     # sales_jan
print(report.suffix)   # .csv
print(report.parent)   # shopeasy/reports
````

## Step 3 · Create folders and files
`mkdir(parents=True, exist_ok=True)` makes the whole chain and does not complain if it already exists. `write_text` creates a file in one line.

````python
project.mkdir(exist_ok=True)

files = {
    "orders.csv": "order_id,amount\n1001,499\n1002,1299\n",
    "logo.png": "(pretend image bytes)",
    "notes.txt": "Call supplier on Monday",
    "summary.csv": "month,total\nJan,1798\n",
    "banner.jpg": "(pretend image bytes)",
}
for name, content in files.items():
    (project / name).write_text(content)

print(sorted(p.name for p in project.iterdir()))
````

## Step 4 · Find files with glob
`glob('*.csv')` finds every CSV in a folder. `rglob` searches sub folders too. `stat().st_size` gives the size in bytes.

````python
for csv_file in sorted(project.glob("*.csv")):
    print(f"{csv_file.name:<12} {csv_file.stat().st_size:>4} bytes")

print()
print(project.joinpath("orders.csv").read_text())
````

## Step 5 · Mini project: tidy the folder
Move each file into a sub folder named after its type. This is a real downloads-folder cleaner in 10 lines.

````python
groups = {".csv": "data", ".png": "images", ".jpg": "images", ".txt": "docs"}

for f in list(project.iterdir()):
    if f.is_file():
        target = project / groups.get(f.suffix, "other")
        target.mkdir(exist_ok=True)
        f.rename(target / f.name)

for folder in sorted(project.iterdir()):
    print(folder.name + "/", sorted(p.name for p in folder.iterdir()))
````

## Step 6 · Environment variables with os
`os.environ` holds settings the system passes to programs. `.get()` with a default avoids errors when one is missing.

````python
os.environ["SHOP_MODE"] = "test"
print(os.environ.get("SHOP_MODE"))
print(os.environ.get("MISSING_SETTING", "default value"))
````
