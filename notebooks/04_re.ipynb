A regular expression is a tiny pattern language for text. Instead of searching for one exact word, you describe a shape, like ten digits, or letters then an at sign then a domain. The re module finds, checks, extracts and replaces anything that matches that shape.

**In this notebook:** You will pull phone numbers and emails out of customer messages, check whether an order code is valid, pull fields out of a log line with named groups, and clean up messy spacing.

## Setup
Nothing to install: this is part of Python's **standard library**, so it works everywhere Python runs.

## Step 1 · findall: get every match
`\d` is a digit, `{5}` means exactly five. `findall` returns a list.

````python
import re

msg = "Call me on 98765-43210 or 91234-56789 after 6pm"
print(re.findall(r"\d{5}-\d{5}", msg))
````

## Step 2 · Extract emails
`[\w.]+` is one or more letters, digits, underscores or dots.

````python
text = "Contact asha.k@shopeasy.in or support@shopeasy.in, not shop@@bad"
emails = re.findall(r"[\w.+-]+@[\w-]+\.[\w.]+", text)
print(emails)
````

## Step 3 · Validate with fullmatch
`fullmatch` checks the whole string fits. Our order codes look like `SE-2026-00042`.

````python
pattern = re.compile(r"SE-\d{4}-\d{5}")
for code in ["SE-2026-00042", "SE-26-42", "se-2026-00042", "SE-2026-000421"]:
    print(f"{code:<16}", "valid" if pattern.fullmatch(code) else "invalid")
````

## Step 4 · Named groups pull out fields
`(?P<name>...)` captures a part and gives it a name.

````python
log = "2026-11-19 18:45:02 ERROR payment failed order=1043 amount=1299"
m = re.search(r"(?P<date>\S+) (?P<time>\S+) (?P<level>[A-Z]+) .* order=(?P<order>\d+) amount=(?P<amount>\d+)", log)
print(m.group("level"), m.group("order"), m.group("amount"))
print(m.groupdict())
````

## Step 5 · Clean text with sub
`sub` replaces matches. `\s+` is any run of spaces, tabs or newlines.

````python
messy = "  Steel   Bottle \t 1 L   (Blue)  "
clean = re.sub(r"\s+", " ", messy).strip()
print(repr(clean))

phone = "+91 98765 43210"
print(re.sub(r"\D", "", phone))      # \D = anything that is NOT a digit
````
