## Setup

## Step 1 · logging instead of print
`force=True` resets Colab's own logging setup. Each line shows time, level and message.

````python
import logging

logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)-8s %(message)s", datefmt="%H:%M:%S", force=True)
log = logging.getLogger("shopeasy")

log.debug("You will NOT see this, level is INFO")
log.info("Store opened")
log.warning("Stock low for item %s", "B-01")
log.error("Payment failed for order %d", 1043)
````

## Step 2 · A function with logging
Log useful facts, not every line.

````python
def apply_discount(amount: float, percent: float) -> float:
    if not 0 <= percent <= 90:
        log.error("Bad discount %s%%", percent)
        raise ValueError("percent must be between 0 and 90")
    final = round(amount * (1 - percent / 100), 2)
    log.info("Discount %s%% on %s -> %s", percent, amount, final)
    return final

apply_discount(1000, 15)
````

## Step 3 · argparse: options for your script
In a notebook, pass a list to `parse_args`. On a laptop, Python reads the real command line.

````python
import argparse

parser = argparse.ArgumentParser(description="Apply a discount")
parser.add_argument("amount", type=float, help="original price")
parser.add_argument("--percent", type=float, default=10, help="discount percent")
parser.add_argument("--verbose", action="store_true", help="show debug logs")

args = parser.parse_args(["1299", "--percent", "20"])
print(args)
print("Final price:", apply_discount(args.amount, args.percent))
parser.print_help()
````

## Step 4 · unittest: prove it works
Each `test_` method checks one behaviour. `exit=False` keeps the notebook running.

````python
import unittest
logging.getLogger("shopeasy").setLevel(logging.CRITICAL)   # quiet logs during tests

class TestDiscount(unittest.TestCase):
    def test_normal(self):
        self.assertEqual(apply_discount(1000, 15), 850)

    def test_zero(self):
        self.assertEqual(apply_discount(499, 0), 499)

    def test_too_big(self):
        with self.assertRaises(ValueError):
            apply_discount(100, 95)

unittest.main(argv=["ignored"], exit=False, verbosity=2)
````

## Step 5 · Run it as a real script
`%%writefile` saves a cell as a file. `!python` runs it in Colab's terminal, the same as on your laptop.

````python
%%writefile discount.py
import argparse

def apply_discount(amount, percent):
    return round(amount * (1 - percent / 100), 2)

if __name__ == "__main__":
    p = argparse.ArgumentParser()
    p.add_argument("amount", type=float)
    p.add_argument("--percent", type=float, default=10)
    a = p.parse_args()
    print(f"Pay Rs.{apply_discount(a.amount, a.percent)}")
````
