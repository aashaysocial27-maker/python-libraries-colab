## Step 1 · Combinations and permutations
`combinations` ignores order (A+B is the same as B+A). `permutations` cares about order.

````python
from itertools import combinations, permutations, product, chain, islice, count, groupby

items = ["bottle", "mat", "lamp"]
print("Bundles of 2:", list(combinations(items, 2)))
print("Podium orders:", len(list(permutations(items, 2))))
````

## Step 2 · product: every variant
`product` is a nested loop in one line.

````python
sizes = ["S", "M", "L"]
colours = ["black", "blue"]
for size, colour in product(sizes, colours):
    print(f"T-shirt {size}-{colour}")
````

## Step 3 · Streams: chain, islice, count
`count()` never ends, so `islice` takes just the first few. `chain` joins lists without copying.

````python
ids = count(start=1001)
print("Next 3 order ids:", list(islice(ids, 3)))
print("Joined:", list(chain(["a", "b"], ["c"], ["d", "e"])))
````

## Step 4 · groupby (sort first!)
Groups neighbours that share a key.

````python
orders = [("Mon", 499), ("Tue", 899), ("Mon", 1299), ("Tue", 199), ("Wed", 999)]
orders.sort(key=lambda o: o[0])
for day, group in groupby(orders, key=lambda o: o[0]):
    amounts = [amt for _, amt in group]
    print(day, amounts, "total", sum(amounts))
````

## Step 5 · lru_cache: remember answers
Without caching, this recursive function repeats work millions of times. With one decorator it is instant.

````python
import time
from functools import lru_cache, partial, reduce

def slow_fib(n):
    return n if n < 2 else slow_fib(n - 1) + slow_fib(n - 2)

@lru_cache(maxsize=None)
def fast_fib(n):
    return n if n < 2 else fast_fib(n - 1) + fast_fib(n - 2)

t = time.perf_counter(); slow_fib(27); slow = time.perf_counter() - t
t = time.perf_counter(); fast_fib(27); fast = time.perf_counter() - t
print(f"slow: {slow:.3f}s  fast: {fast:.6f}s")
print("fib(90) =", fast_fib(90))
````

## Step 6 · partial and reduce
`partial` pre-fills arguments. `reduce` folds a list into one value.

````python
def price_with_tax(amount, rate):
    return round(amount * (1 + rate), 2)

gst18 = partial(price_with_tax, rate=0.18)
print(gst18(499), gst18(1000))

print("Product of 1..6:", reduce(lambda a, b: a * b, range(1, 7)))
````
