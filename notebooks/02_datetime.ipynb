Dates look simple until you try to add thirty days to January thirtieth, or compare a time in Mumbai with one in London. The datetime module does that maths correctly. A datetime is a moment. A timedelta is a length of time. You can add a timedelta to a datetime to move forwards or backwards.

## Step 1 · Today and now
`date` is a calendar day. `datetime` adds the time of day.

````python
from datetime import date, datetime, timedelta

today = date.today()
now = datetime.now()
print("Today:", today)
print("Now:  ", now.strftime("%H:%M:%S"))
print("Year, month, day:", today.year, today.month, today.day)
````

## Step 2 · Date maths with timedelta
Add or subtract a `timedelta`. Subtracting two dates gives a `timedelta` too.

````python
order_day = date(2026, 1, 30)
delivery = order_day + timedelta(days=5)
print("Delivery on:", delivery)          # crosses into February correctly

sale_ends = date(2026, 12, 31)
left = sale_ends - date(2026, 11, 19)
print("Days until sale ends:", left.days)
````

## Step 3 · Format dates for people
`strftime` turns a date into text. `%d` day, `%b` short month, `%A` weekday, `%Y` year.

````python
d = datetime(2026, 11, 19, 18, 45)
print(d.strftime("%d %b %Y"))          # 19 Nov 2026
print(d.strftime("%A, %I:%M %p"))      # Thursday, 06:45 PM
print(d.strftime("%Y-%m-%d"))          # ISO style for files
````

## Step 4 · Read dates from text
`strptime` parses text using the same codes. Each source format needs its own pattern.

````python
raw = ["19/11/2026", "01/02/2026", "25/12/2026"]
parsed = [datetime.strptime(s, "%d/%m/%Y").date() for s in raw]
print(parsed)
print("Earliest:", min(parsed))
print("ISO text back to date:", date.fromisoformat("2026-11-19"))
````

## Step 5 · Time zones
`zoneinfo` (built in since Python 3.9) gives real time zones, including daylight saving.

````python
from zoneinfo import ZoneInfo

placed = datetime(2026, 11, 19, 21, 30, tzinfo=ZoneInfo("Asia/Kolkata"))
print("India:   ", placed.strftime("%d %b %H:%M %Z"))
print("New York:", placed.astimezone(ZoneInfo("America/New_York")).strftime("%d %b %H:%M %Z"))
print("London:  ", placed.astimezone(ZoneInfo("Europe/London")).strftime("%d %b %H:%M %Z"))
````

## Step 6 · Mini project: age calculator
Subtract years, then take one off if the birthday has not happened yet this year.

````python
def age_on(birthday: date, on: date) -> int:
    had_birthday = (on.month, on.day) >= (birthday.month, birthday.day)
    return on.year - birthday.year - (0 if had_birthday else 1)

print(age_on(date(2000, 12, 25), date(2026, 11, 19)))   # 25
print(age_on(date(2000, 3, 1), date(2026, 11, 19)))     # 26
````
