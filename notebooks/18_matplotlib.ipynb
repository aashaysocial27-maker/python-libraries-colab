## Step 1 · Create the ShopEasy data
The same 600 orders used in the pandas lessons.

````python
import numpy as np, pandas as pd

rng = np.random.default_rng(42)
n = 600
orders = pd.DataFrame({
    "order_id": np.arange(1001, 1001 + n),
    "date": pd.Timestamp("2026-01-01") + pd.to_timedelta(rng.integers(0, 90, n), unit="D"),
    "city": rng.choice(["Mumbai", "Delhi", "Pune", "Bengaluru"], n),
    "category": rng.choice(["Books", "Electronics", "Clothing", "Home"], n, p=[0.3, 0.2, 0.3, 0.2]),
    "quantity": rng.integers(1, 6, n),
    "price": rng.choice([199, 499, 999, 1499, 2999], n),
    "rating": rng.choice([1, 2, 3, 4, 5], n, p=[0.05, 0.1, 0.2, 0.35, 0.3]),
})
orders["revenue"] = orders["quantity"] * orders["price"]
orders.to_csv("shop_orders.csv", index=False)
print(orders.shape, "-> saved shop_orders.csv")
orders.head()
````

## Step 2 · Your first line chart
`fig, ax = plt.subplots()` makes one chart. Group revenue by day and plot it.

````python
import matplotlib.pyplot as plt

daily = orders.groupby("date")["revenue"].sum()
fig, ax = plt.subplots(figsize=(9, 3.5))
ax.plot(daily.index, daily.values, color="#2C5A9A", linewidth=1.5)
ax.set_title("ShopEasy daily revenue, Jan to Mar 2026")
ax.set_xlabel("Date")
ax.set_ylabel("Revenue (Rs.)")
ax.grid(alpha=0.3)
plt.show()
````

## Step 3 · Bar chart with labels
`bar_label` writes the value on each bar.

````python
by_city = orders.groupby("city")["revenue"].sum().sort_values()
fig, ax = plt.subplots(figsize=(7, 3.5))
bars = ax.barh(by_city.index, by_city.values / 1000, color="#E0A93E")
ax.bar_label(bars, fmt="%.0fk", padding=4)
ax.set_title("Revenue by city (Rs. thousands)")
ax.spines[["top", "right"]].set_visible(False)
plt.show()
````

## Step 4 · Scatter and histogram
Two charts side by side with `plt.subplots(1, 2)`.

````python
fig, (left, right) = plt.subplots(1, 2, figsize=(10, 3.5))

left.scatter(orders["price"], orders["revenue"], c=orders["quantity"], cmap="viridis", alpha=0.6)
left.set_title("Price vs revenue (colour = quantity)")
left.set_xlabel("Price"); left.set_ylabel("Revenue")

right.hist(orders["revenue"], bins=30, color="#2C5A9A", edgecolor="white")
right.axvline(orders["revenue"].median(), color="red", linestyle="--", label="median")
right.set_title("How big are orders?")
right.legend()
plt.tight_layout()
plt.show()
````

## Step 5 · Mini project: a 2x2 dashboard saved to a file
`savefig` writes a PNG you can put in a report. In Colab, find it in the Files panel on the left.

````python
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
orders.groupby("category")["revenue"].sum().plot.pie(ax=axes[0, 0], autopct="%1.0f%%", ylabel="")
axes[0, 0].set_title("Revenue share by category")

orders.groupby(orders["date"].dt.month)["revenue"].sum().plot.bar(ax=axes[0, 1], color="#2C5A9A")
axes[0, 1].set_title("Revenue by month"); axes[0, 1].set_xlabel("Month")

orders["rating"].value_counts().sort_index().plot.bar(ax=axes[1, 0], color="#1D7A46")
axes[1, 0].set_title("Ratings count")

orders.boxplot(column="revenue", by="category", ax=axes[1, 1])
axes[1, 1].set_title("Order value by category"); fig.suptitle("")

plt.tight_layout()
fig.savefig("shopeasy_dashboard.png", dpi=150)
plt.show()
print("Saved shopeasy_dashboard.png")
````
