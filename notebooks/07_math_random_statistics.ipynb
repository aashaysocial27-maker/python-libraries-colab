## Step 1 · math basics
`ceil` rounds up, `floor` rounds down. Handy for packing and pagination.

````python
import math, random, statistics

items, per_box = 23, 5
print("Boxes needed:", math.ceil(items / per_box))
print("Full boxes:  ", math.floor(items / per_box))
print("sqrt(144) =", math.sqrt(144), "| pi =", round(math.pi, 4))
print("Ways to pick 3 of 10:", math.comb(10, 3))
````

## Step 2 · Random numbers you can repeat
Same seed means same results. Run this cell twice: nothing changes.

````python
random.seed(42)
print([random.randint(1, 6) for _ in range(5)])   # dice rolls
random.seed(42)
print([random.randint(1, 6) for _ in range(5)])   # identical
````

## Step 3 · choice, sample, shuffle
`sample` picks without repeats, which is fair for a lucky draw.

````python
random.seed(7)
customers = ["Asha", "Ravi", "Meera", "John", "Priya", "Arjun", "Zoya", "Kabir"]
print("Lucky winners:", random.sample(customers, 3))
print("Random pick:", random.choice(customers))

shuffled = customers[:]
random.shuffle(shuffled)
group_a, group_b = shuffled[:4], shuffled[4:]
print("A/B test groups:", group_a, group_b)
````

## Step 4 · Quick statistics
Mean vs median: one big sale pulls the mean up but not the median.

````python
week_sales = [12000, 15000, 11000, 14000, 13000, 12500, 61000]   # Sunday had a flash sale
print("Mean:  ", round(statistics.mean(week_sales)))
print("Median:", statistics.median(week_sales))
print("Stdev: ", round(statistics.stdev(week_sales)))
print("Mode of ratings:", statistics.mode([5, 4, 5, 3, 5, 4]))
````

## Step 5 · Mini simulation
Simulate 10,000 days where 3% of 200 visitors buy. Simulation is a powerful way to answer 'what usually happens?'

````python
random.seed(1)
days = [sum(random.random() < 0.03 for _ in range(200)) for _ in range(10_000)]
print("Average orders per day:", statistics.mean(days))
print("Worst day:", min(days), "| Best day:", max(days))
````
