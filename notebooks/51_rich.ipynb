## Step 1 · Colours and pretty printing
Markup in square brackets. `print` from rich formats any object nicely.

````python
from rich import print as rprint
from rich.console import Console
console = Console()

console.print("[bold green]ShopEasy[/] daily report: [bold]Rs.1,45,200[/] revenue, [red]3 failed payments[/]")
rprint({"orders": 238, "top_city": "Pune", "categories": ["Books", "Home", "Electronics"], "growth": 0.12})
console.rule("[bold blue]Sales")
````

## Step 2 · Tables
Columns with alignment and styles; rows added one by one.

````python
from rich.table import Table
table = Table(title="Revenue by city", show_lines=False)
table.add_column("City", style="cyan")
table.add_column("Orders", justify="right")
table.add_column("Revenue", justify="right", style="green")
table.add_column("Trend")
for city, n, rev, trend in [("Pune", 169, 660869, "[green]▲ 8%"), ("Mumbai", 146, 569185, "[red]▼ 3%"), ("Delhi", 137, 533984, "[green]▲ 2%")]:
    table.add_row(city, str(n), f"Rs.{rev:,}", trend)
console.print(table)
````

## Step 3 · Panels and trees
Panels draw a box; trees show nested structure.

````python
from rich.panel import Panel
from rich.tree import Tree
console.print(Panel("Stock low: [bold]Yoga Mat[/] (3 left)\nCourier delay in [bold]Delhi[/]", title="Alerts", border_style="yellow"))
tree = Tree("[bold]Catalogue")
home = tree.add("Home"); home.add("Desk Lamp"); home.add("Table Fan")
fit = tree.add("Fitness"); fit.add("Yoga Mat"); fit.add("Dumbbells")
console.print(tree)
````

## Step 4 · Progress bars
`track` wraps any loop. Works in Colab and in a terminal.

````python
import time
from rich.progress import track
total = 0
for order in track(range(40), description="Processing orders..."):
    time.sleep(0.03)
    total += order
console.print(f"Done. Checksum {total}")
````

## Step 5 · Code and errors
Syntax-highlighted code, and tracebacks that show local variables.

````python
from rich.syntax import Syntax
console.print(Syntax("def total(items):\n    return sum(p * q for p, q in items)", "python", line_numbers=True))

def average_order(orders):
    return sum(orders) / len(orders)
try:
    average_order([])
except ZeroDivisionError:
    console.print_exception(show_locals=True, max_frames=2)
````
