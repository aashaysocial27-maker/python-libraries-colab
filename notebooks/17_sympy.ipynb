Most Python maths works with numbers. SymPy works with symbols, like x and y, the way you do algebra on paper. It can expand and simplify expressions, solve equations exactly, take derivatives and integrals, and print the answers as neat maths. It is a free computer algebra system inside Python.

## Step 1 · Symbols and expressions
Declare symbols, then build, expand, factor and simplify expressions.

````python
import sympy as sp

x, y = sp.symbols("x y")
expr = (x + 2) * (x - 3)
print("Expanded:", sp.expand(expr))
print("Factored:", sp.factor(x**2 - 9))
print("Simplified:", sp.simplify((x**2 - 1) / (x - 1)))
print("Exact sqrt(8):", sp.sqrt(8), "| decimal:", sp.sqrt(8).evalf(6))
````

## Step 2 · Solve equations
`Eq(left, right)` builds an equation. `solve` returns exact answers.

````python
# Break-even: fixed cost 50,000 + 300 per unit = 799 per unit sold
units = sp.symbols("units", positive=True)
print("Break-even units:", sp.solve(sp.Eq(50_000 + 300 * units, 799 * units), units))

print("Roots of x^2 - 5x + 6:", sp.solve(x**2 - 5*x + 6, x))
print("System:", sp.solve([x + y - 10, 2*x - y - 2], [x, y]))
````

## Step 3 · Derivatives: find the best price
Profit peaks where its derivative is zero. Same problem as the SciPy lesson, solved exactly.

````python
p = sp.symbols("p", positive=True)
profit = (p - 300) * 1000 * sp.exp(-p / 800)
d_profit = sp.diff(profit, p)
best = sp.solve(sp.Eq(d_profit, 0), p)
print("Derivative:", sp.simplify(d_profit))
print("Best price:", best, "->", [float(b) for b in best])
````

## Step 4 · Integrals and limits
`integrate` with and without bounds, and `limit`.

````python
print(sp.integrate(3 * x**2 + 2 * x, x))
print(sp.integrate(sp.exp(-x), (x, 0, sp.oo)))
print(sp.limit(sp.sin(x) / x, x, 0))
print("Exact fraction:", sp.Rational(1, 3) + sp.Rational(1, 6))
````

## Step 5 · Pretty maths and fast functions
Colab shows SymPy results as typeset maths. `lambdify` turns a formula into a NumPy function for speed.

````python
import numpy as np
f = sp.lambdify(p, profit, "numpy")
prices = np.array([500, 800, 1100, 1400])
print(dict(zip(prices.tolist(), f(prices).round(0).tolist())))
sp.Integral(3 * x**2, x)       # shows as maths in Colab
````
