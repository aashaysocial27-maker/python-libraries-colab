JSON is the language web services use to swap data. It looks almost exactly like a Python dictionary, with keys, values, lists and nested objects. The json module converts between the two. Dumps turns Python into JSON text. Loads turns JSON text back into Python.

**In this notebook:** You will save the ShopEasy product catalogue to a JSON file, load it back, dig into nested data, and handle a value JSON does not understand, a date.

## Setup
Nothing to install: this is part of Python's **standard library**, so it works everywhere Python runs.

## Step 1 · Python to JSON text
`json.dumps` returns a string. `indent=2` makes it readable.

````python
import json

product = {"id": 101, "name": "Steel Bottle", "price": 499, "in_stock": True, "tags": ["kitchen", "eco"]}
text = json.dumps(product, indent=2)
print(text)
print(type(text))
````

## Step 2 · JSON text back to Python
`json.loads` returns normal Python objects. Note `true` became `True`.

````python
api_reply = '{"order_id": 1001, "paid": true, "items": [{"sku": "B-01", "qty": 2}], "coupon": null}'
data = json.loads(api_reply)
print(data["paid"], data["coupon"])
print("First item qty:", data["items"][0]["qty"])
````

## Step 3 · Save and load a file
`json.dump` writes to an open file. `json.load` reads it back.

````python
catalogue = [
    {"id": 101, "name": "Steel Bottle", "price": 499},
    {"id": 102, "name": "Yoga Mat", "price": 1299},
    {"id": 103, "name": "Desk Lamp", "price": 899},
]
with open("catalogue.json", "w") as f:
    json.dump(catalogue, f, indent=2)

with open("catalogue.json") as f:
    loaded = json.load(f)
print(len(loaded), "products. Cheapest:", min(loaded, key=lambda p: p["price"])["name"])
````

## Step 4 · Values JSON cannot store
Dates are not JSON. Pass `default=str` to convert unknown types to text.

````python
from datetime import date
order = {"id": 1001, "placed": date(2026, 11, 19)}
try:
    json.dumps(order)
except TypeError as e:
    print("Error:", e)
print(json.dumps(order, default=str))
````

## Step 5 · Mini project: update a stock file
Load, change, save. This is how small apps keep settings and data.

````python
with open("catalogue.json") as f:
    items = json.load(f)
for item in items:
    item["price"] = round(item["price"] * 0.9)   # 10% sale
with open("catalogue.json", "w") as f:
    json.dump(items, f, indent=2)
print(open("catalogue.json").read())
````
