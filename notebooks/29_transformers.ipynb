Training a language model from scratch costs a fortune. Hugging Face hosts hundreds of thousands of models that others have already trained, and the transformers library lets you download and use them in one line. The pipeline function picks the right model and handles all the steps for a task, like sentiment, named entities, or answering questions.

**In this notebook:** You will tag ShopEasy reviews as positive or negative, pull names and places out of support messages, sort messages into topics the model was never trained on with zero shot classification, and look inside a tokenizer.

## Setup
Transformers is pre-installed in Colab. Each model downloads the first time you use it (a few hundred MB, about 10 to 60 seconds). A GPU is optional.

## Step 1 · Sentiment analysis
Downloads a small model the first time.

````python
from transformers import pipeline
reviews = [
    "Absolutely love this steel bottle, keeps water cold all day!",
    "The yoga mat smells bad and started tearing after a week. Very disappointed.",
    "Delivery was quick and the desk lamp is bright. Good value for money.",
    "Terrible customer service. My refund still hasn't arrived after 20 days.",
    "Decent headphones, but the battery life is shorter than advertised.",
    "Best purchase this year! The cotton shirts fit perfectly and feel great.",
    "The book arrived with a torn cover, but the story is wonderful.",
    "Not worth the price. The blender stopped working in two days.",
]

sentiment = pipeline("sentiment-analysis", model="distilbert/distilbert-base-uncased-finetuned-sst-2-english")
for review, result in zip(reviews, sentiment(reviews)):
    print(f"{result['label']:<8} {result['score']:.2f}  {review[:60]}")
````

## Step 2 · Named entities
Finds people, organisations and places in text.

````python
ner = pipeline("ner", model="dslim/bert-base-NER", aggregation_strategy="simple")
msg = "Hi, this is Priya Sharma from Pune. My Samsung phone ordered through ShopEasy never reached Mumbai."
for ent in ner(msg):
    print(f"{ent['entity_group']:<5} {ent['word']:<15} {ent['score']:.2f}")
````

## Step 3 · Zero-shot classification
Give your own labels at run time. No training needed.

````python
zero = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
labels = ["delivery", "refund", "product quality", "praise"]
for text in reviews[:4]:
    r = zero(text, candidate_labels=labels)
    print(f"{r['labels'][0]:<16} {r['scores'][0]:.2f}  {text[:55]}")
````

## Step 4 · Inside the tokenizer
Words become tokens, and tokens become numbers.

````python
from transformers import AutoTokenizer
tok = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")
text = "ShopEasy's unbelievably fast delivery!"
print(tok.tokenize(text))
print(tok(text)["input_ids"])
print("Max length this model accepts:", tok.model_max_length)
````

## Step 5 · Fill in the blank
A masked language model guesses the hidden word.

````python
fill = pipeline("fill-mask", model="distilbert/distilbert-base-uncased")
for g in fill("The delivery was very [MASK] and the product works well.")[:3]:
    print(f"{g['token_str']:<10} {g['score']:.2f}")
````
